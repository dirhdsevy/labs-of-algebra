### 1. Класи суміжності та нормальні підгрупи

Розглянемо групу $S_4$ та її підгрупи:<br>
$\quad$ (a) $H_1=\langle (1,2,3,4)\rangle$;<br>
$\quad$ (b) $H_2=\langle (1,3), (1,2,3,4)\rangle$;<br>
$\quad$ (c) $H_3=K_4$.

Для кожної з цих підгруп $H$ виконайте наступні кроки:
1. Знайдіть порядок $H$ та її індекс $[S_4:H]$.
2. Перевірте, чи є підгрупа $H$ нормальною.  % H.is_normal(G)
3. Якщо $H$ є нормальною, то побудуйте факторгрупу $G/H$, її таблицю Келі та визначіть структуру.  % G.quotient(H), .structure_description()
4. Якщо $H$ не є нормальною, то знайдіть для неї ліві та праві класи суміжності.  % G.cosets(H) 


```python
print("=== ЗАВДАННЯ 1 ===")
G = SymmetricGroup(4)
print(f"Група S₄, порядок: {G.order()}")

H1 = G.subgroup([(1,2,3,4)])
print(f"\n(a) H₁ = ⟨(1,2,3,4)⟩")
print(f"Порядок H₁: {H1.order()}")
print(f"Індекс [S₄:H₁]: {G.order() // H1.order()}")
print(f"Чи нормальна: {H1.is_normal(G)}")

if not H1.is_normal(G):
    left_cosets = []
    for g in G:
        coset = set(g * h for h in H1)
        if coset not in left_cosets:
            left_cosets.append(coset)
    print(f"Кількість лівих класів суміжності: {len(left_cosets)}")

H2 = G.subgroup([(1,3), (1,2,3,4)])
print(f"\n(b) H₂ = ⟨(1,3), (1,2,3,4)⟩")
print(f"Порядок H₂: {H2.order()}")
print(f"Індекс [S₄:H₂]: {G.order() // H2.order()}")
print(f"Чи нормальна: {H2.is_normal(G)}")

element1 = G('(1,2)(3,4)')
element2 = G('(1,3)(2,4)')
H3 = G.subgroup([element1, element2])
print(f"\n(c) H₃ = K₄")
print(f"Порядок H₃: {H3.order()}")
print(f"Індекс [S₄:H₃]: {G.order() // H3.order()}")
print(f"Чи нормальна: {H3.is_normal(G)}")

if H3.is_normal(G):
    Q = G.quotient(H3)
    print(f"Факторгрупа S₄/K₄:")
    print(f"Порядок: {Q.order()}")
    print(f"Структура: {Q.structure_description()}")
    print("Приклад таблиці Келі для S₃:")
    S3 = SymmetricGroup(3)
    print(S3.cayley_table())
```

### 2. Нормальні підгрупи в дієдральних групах

Для значень $n=3,4,\ldots,100$ виконайте наступні кроки:
1. Задайте дієдральну групу $D_n$.
2. Визначіть кількість нормальних підгруп в $D_n$.
3. Побудуйте список порядків нормальних підгруп.
4. Визначіть структуру всіх нормальних підгруп.  %  .structure_description()
5. Знайдіть закономірність і виведіть гіпотезу про кількість та структуру нормальних підгруп в $D_n$.
6. Спрогнозуйте скільки нормальних підгруп має група $D_{470448}$ без обчислення. 

```python
print("\n=== ЗАВДАННЯ 2 ===")

def analyze_dihedral(n):
    try:
        D = DihedralGroup(n)
        normal_subs = D.normal_subgroups()
        
        print(f"\nD_{n} (порядок {2*n}):")
        print(f"Кількість нормальних підгруп: {len(normal_subs)}")
        
        orders = [sg.order() for sg in normal_subs]
        
        print(f"Порядки нормальних підгруп: {sorted(orders)}")
        
        if n <= 8:
            structures = []
            for sg in normal_subs:
                try:
                    structures.append(sg.structure_description())
                except:
                    structures.append("?")
            print(f"Структури: {structures}")
        
        return len(normal_subs), orders
    except Exception as e:
        print(f"Помилка для D_{n}: {e}")
        return 0, []

print("Аналіз нормальних підгруп дієдральних груп:")
for n in range(3, 9):
    analyze_dihedral(n)

print("\n=== ЗАКОНОМІРНОСТІ ===")
print("Для непарних n: зазвичай 3 нормальні підгрупи - {1}, C_n, D_n")
print("Для парних n: більше нормальних підгруп через додаткові підгрупи")
```

### 3. Проекції групи $SL_2(\mathbb{Z}_m)$ на $SL_2(\mathbb{Z}_n)$ 

Для значень $m=6,8,20,30,100$ задайте групу $SL_2(\mathbb{Z}_m)$.
Для кожного дільника $n$ числа $m$:

1. Задайте групу $SL_2(\mathbb{Z}_n)$.
2. Задайте гомоморфізм з групи $SL_2(\mathbb{Z}_m)$ в групу $SL_2(\mathbb{Z}_n)$, який редукує матриці mod n. % M.change_ring(Integers(n))
3. Перевірте, чи є гомоморфізм сюр'єктивним.<br>
4. Знайдіть порядок ядра гомоморфізму та порівняйте з $|SL_2(\mathbb{Z}_m)| / |SL_2(\mathbb{Z}_n)|$.<br>



```python
print("\n=== ЗАВДАННЯ 3 ===")

def analyze_sl2_simple(m):
    print(f"\nАналіз для m = {m}")
    
    def sl2_order(n):
        """Обчислює порядок SL₂(ℤₙ)"""
        if n == 1:
            return 1
        result = 1
        for p, e in factor(n):
            if p == 2:
                order_p = 2**(3*e - 2) * 3
            else:
                order_p = p**(3*e - 2) * (p**2 - 1)
            result *= order_p
        return result
    
    divisors_m = m.divisors()
    print(f"Дільники m: {divisors_m}")
    
    order_m = sl2_order(m)
    print(f"|SL₂(ℤₘ)| = {order_m}")
    
    for n in divisors_m:
        if n == 1:
            continue
            
        order_n = sl2_order(n)
        expected_kernel = order_m / order_n
        
        print(f"  n = {n}:")
        print(f"    |SL₂(ℤₙ)| = {order_n}")
        print(f"    Очікуваний порядок ядра: {expected_kernel}")

for m in [6, 8, 12]:
    analyze_sl2_simple(m)
```
