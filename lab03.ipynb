### 1. Властивості циклічних груп

Для $n=100,173,530,746,911,1024$ розглянемо циклічну групу $\mathbb{Z}_n$ (або $C_n$):
1. Для кожного дільника $d$ порядку $n$:<br>
   1.1 Знайдіть кількість елементів порядку $d$ в групі $\mathbb{Z}_n$ (або $C_n$).<br>
   1.2 Порівняйте з $\phi(d)$.
2. Знайдіть кількість підгруп в групі $\mathbb{Z}_n$ (або $C_n$).<br>
   Порівняйте з $\tau(n)$.

Проаналізуйте результати та дайте відповіді на запитання:
1. Скільки елементів порядку $k$ в циклічній групі порядку $n$?
2. Cкільки підгруп в циклічній групі порядку $n$?

```python
n_values = [100, 173, 530, 746, 911, 1024]

for n in n_values:
    print(f"\n--- Циклічна група Z_{n} ---")
    G = CyclicPermutationGroup(n)
    
    divisors = n.divisors()
    print(f"Дільники {n}: {divisors}")
    
    print("\nПорівняння кількості елементів порядку d з φ(d):")
    for d in divisors:
        if d == 1:
            continue
            
        count_order_d = sum(1 for g in G if g.order() == d)
        euler_p = euler_phi(d)

        print(f"d = {d}: елементів порядку d = {count_order_d}, φ({d}) = {euler_p}")

    subgroups = G.subgroups()
    tau_n = len(divisors)
    print(f"\nКількість підгруп: {len(subgroups)}")
    print(f"τ({n}) = {tau_n}")
    print(f"Збігаються: {len(subgroups) == tau_n}")

print("\n" + "="*60)
print("ВИСНОВКИ:")
print("1. Кількість елементів порядку k в циклічній групі порядку n дорівнює φ(k), якщо k ділить n, інакше 0")
print("2. Кількість підгруп в циклічній групі порядку n дорівнює τ(n)")
```

### 2. Порядки елементів в дієдральних групах

Для $n=3,4,...,20$ розглянемо дієдральну групу $D_n$:   
1. Знайдіть кількість елементів порядку $d=2$ в групі $D_n$.
2. Для кожного дільника $d>2$ порядку $|D_n|=2n$:<br>
   Знайдіть кількість елементів порядку $d$ в групі $D_n$.

Проаналізуйте результати та дайте відповіді на запитання:
1. Скільки елементів порядку $d=2$ в дієдральній групі $D_n$?
2. Скільки елементів порядку $d$ в дієдральній групі $D_n$?
3. Чи для кожного дільника $d<2n$ порядку групи $D_n$ існує елемент порядку $d$?

```python
for n in range(3, 21):
    Dn = DihedralGroup(n)
    print(f"\n--- Дієдральна група D_{n} (порядок {2*n}) ---")
    
    elements_order_2 = [g for g in Dn if g.order() == 2]
    print(f"Кількість елементів порядку 2: {len(elements_order_2)}")
    
    divisors_gt_2 = [d for d in (2*n).divisors() if d > 2]
    for d in divisors_gt_2:
        elements_order_d = [g for g in Dn if g.order() == d]
        if elements_order_d:
            print(f"Елементів порядку {d}: {len(elements_order_d)}")

print("\n" + "="*60)
print("ВИСНОВКИ:")
print("1. В D_n кількість елементів порядку 2: n + 1 (при парному n) або n (при непарному n)")
print("2. Елементи порядку d>2 існують лише якщо d ділить n, і їх кількість φ(d)")
print("3. Не для кожного дільника d<2n існує елемент порядку d")
```

### 5\*. Групи $\mathbb{Z}_n^{*}$

1. Для кожного $n=2,3,4,\ldots,1000$ надрукуйте: <br>
       1.1. Розклад $n$ у добуток простих.<br>
       1.2. Відповідь, чи є група $\mathbb{Z}_n^{*}$ циклічною.<br>
       1.3. Якщо $\mathbb{Z}_n^{*}$ є циклічною, то знайти найменше $a\in\mathbb{N}$ таке, що $\overline{a}$ є твірним групи.
3. Сформулюйте гіпотезу про те, для яких $n\in\mathbb{N}$ група $\mathbb{Z}_n^{*}$ є циклічною.

__Ваша гіпотеза__:  Текст тут


```python
print("\n\n5*. Групи Z_n*")
print("-" * 40)

cyclic_count = 0
for n in range(2, 101):
    factors = factor(n)
    factors_str = str(factors)
    
    is_cyclic = False
    primitive_root = None
    
    if n == 2 or n == 4:
        is_cyclic = True
    elif is_prime(n):
        is_cyclic = True
    elif len(factors) == 2 and factors[0][0] == 2 and factors[0][1] == 1 and is_prime(factors[1][0]):
        is_cyclic = True
    elif len(factors) == 1 and factors[0][0] != 2 and factors[0][1] >= 1:
        is_cyclic = True
    
    if is_cyclic:
        cyclic_count += 1
        for a in range(1, n):
            if gcd(a, n) == 1:
                if Mod(a, n).multiplicative_order() == euler_phi(n):
                    primitive_root = a
                    break
    
    status = "Так" if is_cyclic else "Ні"
    root_str = str(primitive_root) if primitive_root else "-"
    print(f"{n:2} | {factors_str:15} | {status:9} | {root_str}")

print(f"\nЗ {99} груп Z_n* циклічних: {cyclic_count}")

print("\n" + "="*60)
print("ГІПОТЕЗА:")
print("Група Z_n* є циклічною тоді і тільки тоді, коли:")
print("n = 2, 4, p^k або 2p^k, де p - непарне просте число, k ≥ 1")
```
